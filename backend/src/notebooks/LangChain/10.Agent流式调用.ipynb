```python
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model
from langchain_core.messages import HumanMessage

llm = init_chat_model("deepseek-v4-flash", extra_body={"thinking": {"type": "disabled"}})
agent = create_agent(llm)
messages = {"messages": [HumanMessage("你是科幻小说作家，帮我创造月球首都城市背景。")]}
# response = agent.invoke(messages)
# rich_print(response)
stream = agent.stream_events(input=messages, version="v3")
for message in stream.messages:
    for chunk in message.text:
        print(chunk, end="", flush=True)
```
