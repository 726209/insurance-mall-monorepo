安装LangChain依赖、以及LangChain对模型的支持依赖
uv install langchain langchain-deepseek

```python
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain.tools import tool

# 1.加载环境变量
load_dotenv(override=True)  # 读取.env文件，加入系统环境变量中


# 2.定义工具，就是定义函数，给函数加装饰器 @tool
@tool
def get_weather(location: str):
    """查询指定location的天气"""
    return f"Current weather in {location} is sunny"


@tool
def square_root(x: float) -> float:
    """求x的平方根"""
    return x**0.5


# 3.定义Agent，绑定模型和工具，用函数create_agent
agent = create_agent(model="deepseek-v4-flash", tools=[get_weather, square_root])

# 4.调用Agent，用invoke函数，传入一个Dict，字典要有messages属性，里面是消息的列表
response = agent.invoke({"messages": [{"role": "user", "content": "杭州天气如何？"}]})
print(response)  #
```
