```python
from langchain.chat_models import init_chat_model
from langchain_core.messages import HumanMessage
from pydantic import BaseModel, Field
from rich import print as rich_print

llm = init_chat_model("deepseek-v4-flash", extra_body={"thinking": {"type": "disabled"}})


# 定义 BaseModel 类，用来约束模型的输出，同时封装模型要输出的数据
class CapitalInfo(BaseModel):
    name: str = Field(description="首都名字")
    location: str = Field(description="首都位置")
    vibe: str = Field(description="首都氛围")
    economy: str = Field(description="首都经济来源")


# 将 BaseModel 类与模型绑定，要求输出结构化结果
llm_with_structured_output = llm.with_structured_output(schema=CapitalInfo)
# 调用模型
response = llm_with_structured_output.invoke(
    [HumanMessage("你是科幻小说作家，帮我创造月球首都城市背景。")]
)
rich_print(response)
```
